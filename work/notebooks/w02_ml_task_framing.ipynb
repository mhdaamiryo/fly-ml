## 3. Success metric

*One metric you can defend. What number means 'good'?*

**Selected Metric: Precision@50**

Precision@50 measures how many of the top 50 webpages recommended by our ranking system are actually relevant for content refreshing.

It is calculated as:

Precision@50 = Relevant pages in the top 50 / 50

For example, if the system recommends 50 webpages and 40 of them genuinely need content refreshing, the Precision@50 is:

40 / 50 = 0.80 (80%)

A Precision@50 score of 80% would mean that 40 out of the 50 recommended webpages are relevant.

This metric is suitable because the content team has limited time and resources. We want the highest-priority recommendations to be useful and actionable rather than overwhelming the team with irrelevant pages.

Therefore, our initial target is a Precision@50 of at least 80%. This is a proposed goal, not an achieved result.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
relevant_pages = 40
k = 50

precision_at_50 = relevant_pages / k

print("Precision@50:", precision_at_50)
print("Precision percentage:", precision_at_50 * 100, "%")
```

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

A simple fixed rule could be: "If a webpage has not been updated for more than 365 days, recommend refreshing it."

However, this rule is not always reliable. An old webpage may still receive good search traffic, while a recently updated webpage may already be experiencing declining performance.

Content performance depends on several interacting factors, including content age, impressions, clicks, CTR, average search position, content type, and user engagement.

Machine Learning can analyse these multiple features together and identify patterns that are difficult to capture using a single fixed condition.

For example, it can identify webpages with declining search impressions even if they were updated recently.

The output will help the content team prioritize webpages for manual review and decide which articles should be refreshed first.

Therefore, ML is useful as a decision-support tool because it can consider multiple signals instead of relying entirely on predefined rules. However, its recommendations should be evaluated against a simple rule-based baseline before claiming that it performs better.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
# ============================================
# SECTION 5: WHY ML BEATS A FIXED RULE
# ============================================

# Example of a simple fixed rule
fixed_rule = "Refresh if content age > 365 days"

print("Fixed Rule:")
print(fixed_rule)

# Example webpages
example_pages = pd.DataFrame({
    "page": ["Article A", "Article B", "Article C", "Article D"],
    "content_age_days": [800, 100, 500, 200],
    "impressions_90d": [15000, 1000, 500, 12000],
    "ctr": [5.5, 0.8, 0.5, 6.0]
})

# Apply the fixed rule
example_pages["fixed_rule_recommendation"] = (
    example_pages["content_age_days"] > 365
)

print("\nExample webpage analysis:")
display(example_pages)

print("\nObservation:")
print("Article A is old but has strong search performance.")
print("Article B is relatively new but has low search performance.")
print("A single age-based rule cannot capture every situation.")

print("\nML Advantage:")
print("ML can analyse multiple features together to identify")
print("more useful content refresh priorities.")
```
